### 费用估计
这是在上一个实验室里研发出来的。我们在这里还会用到它。

```python
#Function to calculate the cost
def compute_cost(x, y, w, b):
   
    m = x.shape[0] 
    cost = 0
    
    for i in range(m):
        f_wb = w * x[i] + b
        cost = cost + (f_wb - y[i])**2
    total_cost = 1 / (2 * m) * cost

    return total_cost
```

## 梯度下降概述
在本课程中，你已经开发了一个预测 $f_{w,b}(x^{(i)})$ 的线性模型:

$$f_{w,b}(x^{(i)}) = wx^{(i)} + b \qquad (1)$$

在训练过程中，你利用输入的训练数据来拟合参数 $w$,$b$ 通过最小化预测值 $f_{w,b}(x^{(i)})$ 与实际数据 $y^{(i)}$ 之间的误差度量。

这个度量被称为成本函数  $J(w,b)$。在训练时，你计算所有训练样本 $x^{(i)},y^{(i)}$ 上的成本：


$$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \qquad (2)$$


在课程中，梯度下降算法被描述为：

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline
\;  w &= w -  \alpha \frac{\partial J(w,b)}{\partial w} \qquad (3)  \; \newline 
 b &= b -  \alpha \frac{\partial J(w,b)}{\partial b}  \newline \rbrace
\end{align*}$$
其中，参数 $w$ 和 $b$ 被同时更新。梯度定义为：
$$
\begin{align*}
\frac{\partial J(w,b)}{\partial w}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})x^{(i)} \qquad (4)\\
  \frac{\partial J(w,b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)}) \qquad (5)\\
\end{align*}
$$

这里的 *同时* 意味着你在更新任何参数之前，先计算所有参数的偏导数。

## 实施梯度下降算法
针对一个特征应用实现梯度下降算法。您将需要三个函数。
- `compute_gradient` 实现上述方程 (4) 和 (5) 的内容。
- `compute_cost` 实现上述方程 (2) 的内容 (来自上一次实验的代码)
- `gradient_descent`, 使用 compute_gradient 和 compute_cost

命名约定:
- 包含偏导数的Python变量命名遵循以下模式：$\frac{\partial J(w,b)}{\partial b}$  将命名为 `dj_db`.
- w.r.t 表示”关于“, 例如 $J(wb)$ 关于 $b$ 的偏导数。


<a name="toc_40291_2.3"></a>
### compute_gradient
<a name='ex-01'></a>
`compute_gradient`  实现上述公式 (4) 和 (5) ，并返回 $\frac{\partial J(w,b)}{\partial w}$ 和 $\frac{\partial J(w,b)}{\partial b}$。 嵌入的注释描述了操作过程。​

```python
def compute_gradient(x, y, w, b): 
    """
    计算线性回归的梯度
    参数:
      x (ndarray (m,)): 数据，m个样本
      y (ndarray (m,)): 目标值
      w,b (scalar)    : 模型参数
    返回
      dj_dw (scalar): 成本关于参数 w 的梯度
      dj_db (scalar): 成本关于参数 b 的梯度
     """
    
    # 训练样本的数量
    m = x.shape[0]    
    dj_dw = 0
    dj_db = 0
    
    for i in range(m):  
        f_wb = w * x[i] + b 
        dj_dw_i = (f_wb - y[i]) * x[i] 
        dj_db_i = f_wb - y[i] 
        dj_db += dj_db_i
        dj_dw += dj_dw_i 
    dj_dw = dj_dw / m 
    dj_db = dj_db / m 
        
    return dj_dw, dj_db
```

请注意，梯度指向远离最小值的方向。回顾上面的公式 (3)： 缩放后的梯度从 $w$ 或 $b$ 的当前值中减去。这使参数向减少成本的方向移动。

###  梯度下降

现在梯度可以计算了，上面公式 (3) 描述的梯度下降可以在下面的 `gradient_descent` 函数中实现。实现细节在注释中描述。接下来，你将使用这个函数在训练数据上找到 
$w$ 和 $b$的最优值。

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters, cost_function, gradient_function): 
    """
    执行梯度下降以拟合 w,b。通过采取 num_iters 次梯度步骤来更新 w,b，学习率为 alpha。
    
    Args:
      x (ndarray (m,))  : 数据，包含 m 个样本 
      y (ndarray (m,))  : 目标值
      w_in,b_in (scalar): 模型参数的初始值
      alpha (float)     : 学习率
      num_iters (int)   : 运行梯度下降的迭代次数
      cost_function     : 用于计算成本的函数
      gradient_function : 用于计算梯度的函数
      
    Returns:
      w (scalar): 运行梯度下降后更新的参数值
      b (scalar): 运行梯度下降后更新的参数值
      J_history (List): 成本值的历史记录
      p_history (list): 参数 [w,b] 的历史记录
      """
    
    w = copy.deepcopy(w_in) # avoid modifying global w_in
    # 一个用于存储每次迭代的成本 J 和 w 的数组，主要用于后续绘图。
    J_history = []
    p_history = []
    b = b_in
    w = w_in
    
    for i in range(num_iters):
        # 使用 gradient_function 计算梯度并更新参数。
        dj_dw, dj_db = gradient_function(x, y, w , b)     

        # 使用上面的方程 (3) 更新参数。
        b = b - alpha * dj_db                            
        w = w - alpha * dj_dw                            

        # 在每次迭代中保存成本 J
        if i<100000:      # 防止资源耗尽。
            J_history.append( cost_function(x, y, w , b))
            p_history.append([w,b])
        # 每间隔10次迭代打印一次成本，如果总迭代次数小于10次，则每次迭代都打印。
        if i% math.ceil(num_iters/10) == 0:
            print(f"Iteration {i:4}: Cost {J_history[-1]:0.2e} ",
                  f"dj_dw: {dj_dw: 0.3e}, dj_db: {dj_db: 0.3e}  ",
                  f"w: {w: 0.3e}, b:{b: 0.5e}")
 
    return w, b, J_history, p_history # 返回 w 以及用于绘图的 J 和 w 的历史记录。
```

```python
# 初始化参数
w_init = 0
b_init = 0
# 一些梯度下降的设置
iterations = 10000
tmp_alpha = 1.0e-2
# 运行梯度下降函数
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha, 
                                                    iterations, compute_cost, compute_gradient)
print(f"(w,b) found by gradient descent: ({w_final:8.4f},{b_final:8.4f})")
```

### 增加学习率

在讲座中，关于学习率 $\alpha$ 在公式(3)中的适当取值进行了讨论。$\alpha$ 越大，梯度下降收敛到解的速度就越快。

但是，如果 $\alpha$ 过大，梯度下降将会发散。上面你看到了一个收敛良好的例子。

让我们尝试增大 $\alpha$ 的值，看看会发生什么：

```python
# initialize parameters
w_init = 0
b_init = 0
# 设置 alpha 为更大的值
iterations = 10
tmp_alpha = 8.0e-1
# 运行
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha, 
                                                    iterations, compute_cost, compute_gradient)
```

在上面, $w$ 和 $b$ 在正值和负值之间来回振荡，并且绝对值随着每次迭代而增加。 

此外，每次迭代中 $\frac{\partial J(w,b)}{\partial w}$ 会改变符号，成本在增加而不是减少。 这是一个明确的迹象，表明 *学习率过大*，解正在发散。

让我们用图来可视化这一点。

```python
plt_divergence(p_hist, J_hist,x_train, y_train)
plt.show()
```

在上面，左图显示了梯度下降最初几步中 $w$ 的变化过程。$w$ 在正值和负值之间振荡，成本迅速增长。梯度下降同时在 $w$ 和 $b$ 上操作，因此需要右边的3D图才能看到完整的画面。​
